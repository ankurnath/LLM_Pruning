```python
def _latex_escape(s: str) -> str:
    # Escape LaTeX special chars in simple labels
    return (s.replace('\\', r'\textbackslash{}')
             .replace('_', r'\_')
             .replace('%', r'\%')
             .replace('&', r'\&')
             .replace('#', r'\#')
             .replace('{', r'\{')
             .replace('}', r'\}')
             .replace('$', r'\$'))

def _fmt4(x):
    if x is None:
        return r'--'
    try:
        return f"{float(x):.4f}"
    except Exception:
        return r'--'

def generate_latex_table(data_dict, problem, algorithms, datasets=None, caption=None, label=None):
    """
    data_dict: your nested dict (Problem -> Algorithm -> Dataset -> {P_r,P_g,C})
    problem: str, e.g. "Maximum Coverage"
    algorithms: list[str], e.g. ["QS", "SS", "GNN", "LLM"]
    datasets: optional list[str] to control row order. If None, inferred from data.
    caption/label: optional LaTeX caption/label.
    """
    if problem not in data_dict:
        raise ValueError(f"Problem '{problem}' not found in data_dict")

    # Collect all datasets across chosen algorithms (preserve user order if provided)
    algo_blocks = {a: data_dict[problem].get(a, {}) for a in algorithms}
    if datasets is None:
        # union of keys while preserving insertion order by algorithms
        seen = []
        for a in algorithms:
            for d in algo_blocks.get(a, {}).keys():
                if d not in seen:
                    seen.append(d)
        datasets = seen

    # Build column spec: 1 (dataset) + 3 per algorithm
    col_spec = "l" + "ccc" * len(algorithms)

    # Header lines
    # Top header: Dataset + multicolumn per algorithm
    top_header = [r"\textbf{Dataset}"]
    for alg in algorithms:
        # Escape algorithm name; also show * if present (escape for LaTeX)
        alg_label = _latex_escape(alg).replace('*', r'\*')
        top_header.append(rf"\multicolumn{{3}}{{c}}{{\textbf{{{alg_label}}}}}")
    top_header = " & ".join(top_header) + r" \\"

    # Second header: three metric columns per algorithm
    sub_header = [""]  # first column is the dataset label
    for _ in algorithms:
        sub_header += [r"\textbf{$P_r$}", r"\textbf{$P_g$}", r"\textbf{$C$}"]
    sub_header = " & ".join(sub_header) + r" \\"

    # Rows
    lines = []
    for ds in datasets:
        row = [_latex_escape(ds)]
        for alg in algorithms:
            m = algo_blocks.get(alg, {}).get(ds, {})
            row += [_fmt4(m.get("P_r")), _fmt4(m.get("P_g")), _fmt4(m.get("C"))]
        lines.append(" & ".join(row) + r" \\")

    # Assemble LaTeX table
    caption = caption or rf"{_latex_escape(problem)} results across datasets"
    label = label or f"tab:{problem.lower().replace(' ', '_')}_algos"

    table = []
    table.append(r"\begin{table}[h]")
    table.append(r"\centering")
    # table.append(rf"\begin{tabular}{{{col_spec}}}")
    table.append(rf"\begin{{tabular}}{{{col_spec}}}")
    table.append(r"\toprule")
    table.append(top_header)
    table.append(sub_header)
    table.append(r"\midrule")
    table.extend(lines)
    table.append(r"\bottomrule")
    table.append(r"\end{tabular}")
    table.append(rf"\caption{{{caption}}}")
    table.append(rf"\label{{{label}}}")
    table.append(r"\end{table}")

    return "\n".join(table)
```

```python
problem = "Maximum Coverage"
algorithms = ["QS","Lense*", "LLM"]  # pick any subset/order you want
datasets = ['Facebook', 'DBLP', 'Deezer', 'Slashdot', 'Wiki', 'Twitter', 'YouTube', 'Skitter']  # optional

latex_str = generate_latex_table(data_dict, problem, algorithms, datasets)
print(latex_str)
```

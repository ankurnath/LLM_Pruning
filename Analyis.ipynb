```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

for problem in [
                'Maximum Coverage',
                'Maximum Cut',
                'Influence Maximization',
                'Maximum Coverage Weighted',
                'Maximum Cut Weighted',
                'Influence Maximization Weighted'
                ]:
    datasets = ['Facebook', 'DBLP', 'Deezer', 'Slashdot', 'Wiki', 'Twitter', 'YouTube','Skitter']

    try:
        data_dict[problem]['LLM2Prune'] = {}
    except:
        data_dict[problem] = {}
        data_dict[problem]['LLM2Prune'] = {}


    for dataset in datasets:
        try:
            results = load_from_pickle(f'{problem}/{dataset}/results/{dataset}.pkl')

            ratio = float(results['Ratio'].iloc[0])
            size = float(results['Size Reduction'].iloc[0])
            combined = ratio * size

            data_dict[problem]['LLM2Prune'][dataset] = {
                "P_r": round(ratio, 4),
                "P_g": round(size, 4),
                "C": round(combined, 4)
            }
        except Exception as e:
            pass

def plot_algo_comparison(data_dict, problems, algos_to_plot, figsize=(12, 6)):
    for problem in problems:
        rows = []
        for algo, datasets in data_dict[problem].items():
            for dataset, vals in datasets.items():
                if vals["C"] is not None:
                    rows.append({"Algorithm": algo, "Dataset": dataset, "C": vals["C"]})
        df = pd.DataFrame(rows)
        df_sub = df[df["Algorithm"].isin(algos_to_plot)]
        plt.figure(figsize=figsize)
        sns.barplot(data=df_sub, x="Dataset", y="C", hue="Algorithm")
        plt.title(f"{problem} – Combined Metric (C)")
        plt.ylabel("Combined Metric (C)")
        plt.xticks(rotation=30)
        plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
        plt.tight_layout()
        plt.show()


plot_algo_comparison (data_dict= data_dict, problems = [
                                        'Maximum Coverage Weighted',
                                        'Maximum Cut Weighted',
                                        'Influence Maximization Weighted',
                                        ],
                      algos_to_plot=["LLM2Prune", "QuickPrune"])
```

```python
def _latex_escape(s: str) -> str:
    # Escape LaTeX special chars in simple labels
    return (s.replace('\\', r'\textbackslash{}')
             .replace('_', r'\_')
             .replace('%', r'\%')
             .replace('&', r'\&')
             .replace('#', r'\#')
             .replace('{', r'\{')
             .replace('}', r'\}')
             .replace('$', r'\$'))

def _fmt4(x):
    if x is None:
        return r'--'
    try:
        return f"{float(x):.4f}"
    except Exception:
        return r'--'



```

```python
# from matplotlib.table import table


def generate_latex_table_multi(data_dict, problems, algorithms, datasets=None, caption=None, label=None):
    """
    data_dict: nested dict (Problem -> Algorithm -> Dataset -> {P_r,P_g,C})
    problems: list[str]
    algorithms: list[str]
    datasets: optional list[str] (applied to each problem)
    caption/label: optional LaTeX caption/label.
    """

    # Column specification: left col + vertical bars separating algorithm blocks
    col_spec = "l|" + "|".join(["ccc"] * len(algorithms))

    # Top header
    top_header = [r"\textbf{Dataset}"]
    for idx, alg in enumerate(algorithms):
        alg_label = _latex_escape(alg).replace('*', r'\*')
        # add "|" only between algorithm blocks
        end_bar = "|" if idx < len(algorithms) - 1 else ""
        top_header.append(rf"\multicolumn{{3}}{{c{end_bar}}}{{\textsc{{{alg_label}}}}}")
    top_header = " & ".join(top_header) + r" \\"

    # Sub-header
    sub_header = [""]
    for _ in algorithms:
        sub_header += [r"\textbf{$P_r$}", r"\textbf{$P_g$}", r"\textbf{$C$}"]
    sub_header = " & ".join(sub_header) + r" \\"

    # light colors for ranks
    rank_colors = {1: "lightred", 2: "lightblue", 3: "lightgreen"}

    lines = []
    for prob in problems:
        if prob not in data_dict:
            continue
        # centered problem row
        lines.append(
            rf"\multicolumn{{{1+3*len(algorithms)}}}{{c}}{{\textbf{{{_latex_escape(prob)}}}}} \\"
        )

        lines.append(
            rf" \midrule"
        )
        algo_blocks = {a: data_dict[prob].get(a, {}) for a in algorithms}

        if datasets is None:
            seen = []
            for a in algorithms:
                for d in algo_blocks.get(a, {}).keys():
                    if d not in seen:
                        seen.append(d)
            prob_datasets = seen
        else:
            prob_datasets = datasets

        for ds in prob_datasets:
            # collect C values
            C_vals = []
            for alg in algorithms:
                m = algo_blocks.get(alg, {}).get(ds, {})
                try:
                    C_vals.append(float(m.get("C")))
                except:
                    C_vals.append(None)

            # rank values
            ranked = {}
            valid = [(i, c) for i, c in enumerate(C_vals) if c is not None]
            sorted_vals = sorted(valid, key=lambda x: x[1], reverse=True)
            for rank, (i, c) in enumerate(sorted_vals, start=1):
                if rank <= 3:
                    ranked[i] = rank

            row = [_latex_escape(ds)]
            for idx, alg in enumerate(algorithms):
                m = algo_blocks.get(alg, {}).get(ds, {})
                pr = _fmt4(m.get("P_r"))
                pg = _fmt4(m.get("P_g"))
                cval = _fmt4(m.get("C"))
                if idx in ranked:
                    color = rank_colors[ranked[idx]]
                    cval = rf"\textbf{{\textcolor{{{color}}}{{{cval}}}}}"
                row += [pr, pg, cval]
            lines.append(" & ".join(row) + r" \\")
        lines.append(r"\midrule")

    caption = caption or "Results across problems and datasets"
    label = label or "tab:multi_problem_algos"

    table = []
    table.append(r"\begin{table*}[h]")
    table.append(r"\centering")
    table.append(rf"\caption{{{caption}}}")
    table.append(r"\vspace{1em}")
    table.append(rf"\label{{{label}}}")
    table.append(r"\resizebox{\textwidth}{!}{%")  # start resize
    table.append(rf"\begin{{tabular}}{{{col_spec}}}")
    table.append(r"\toprule")
    table.append(top_header)
    table.append(sub_header)
    table.append(r"\midrule")
    table.extend(lines)
    table.append(r"\bottomrule")
    table.append(r"\end{tabular}")
    table.append(r"}")  # end resize
    table.append(r"\end{table*}")


    return "\n".join(table)


problems = [ "Maximum Coverage","Maximum Cut","Influence Maximization"]
algorithms = ["QuickPrune","SS","LLM2Prune","GCOMB-P","LeNSE","COMBHelper"]
datasets = ["Facebook", "DBLP", "Deezer", "Slashdot", "Wiki", "Twitter", "YouTube", "Skitter"]


# caption = r"Comparison of pruning algorithms for size constraint experiments (best combined metric in bold); *Values as reported in  \cite{ireland2022lense} and ``--" denotes no reasonable result is achieved by the corresponding algorithm under the time constraint."

# caption = "Comparison of pruning algorithms for size-constraint experiments. Highlighted are the results ranked \textbf{\textcolor{lightred}{first}}, \textbf{\textcolor{lightblue}{second}} , and \textbf{\textcolor{lightgreen}{third}}. *Values are reported from \cite{ireland2022lense,pmlr-v258-nath25a}. “--” indicates that the corresponding algorithm obtained no reasonable result.

caption = "Comparison of pruning algorithms for size-constraint experiments. The results ranked \textbf{\textcolor{lightred}{first}}, \textbf{\textcolor{lightblue}{second}}, and \textbf{\textcolor{lightgreen}{third}} are highlighted. *Values are reported from \cite{ireland2022lense,pmlr-v258-nath25a}. ``--'' indicates that the corresponding algorithm did not obtain a reasonable result."


latex_str = generate_latex_table_multi(data_dict, problems, algorithms, datasets, caption= caption)
print(latex_str)
```

```python
import matplotlib.pyplot as plt
import numpy as np

from matplotlib.patches import Circle, RegularPolygon
from matplotlib.path import Path
from matplotlib.projections import register_projection
from matplotlib.projections.polar import PolarAxes
from matplotlib.spines import Spine
from matplotlib.transforms import Affine2D
from matplotlib.lines import Line2D
from matplotlib.patches import Patch

# --- palette ---
palette = {
    "SS": "#6A5ACD",
    "QuickPrune": "#FF6347",
    "LLM2Prune": "#FFB000",
    "LeNSE": "#20B2AA",
    "GCOMB-P": "#32CD32",
    "COMBHelper": "#FF69B4",
}

# ---------- radar setup ----------
def radar_factory(num_vars, frame='circle'):
    theta = np.linspace(0, 2*np.pi, num_vars, endpoint=False)

    class RadarTransform(PolarAxes.PolarTransform):
        def transform_path_non_affine(self, path):
            if path._interpolation_steps > 1:
                path = path.interpolated(num_vars)
            return Path(self.transform(path.vertices), path.codes)

    class RadarAxes(PolarAxes):
        name = 'radar'
        PolarTransform = RadarTransform
        def __init__(self, *a, **k):
            super().__init__(*a, **k)
            self.set_theta_zero_location('N')  # we'll offset per-axes

        def fill(self, *a, closed=True, **k): return super().fill(closed=closed, *a, **k)
        def plot(self, *a, **k):
            lines = super().plot(*a, **k)
            for ln in lines:
                x, y = ln.get_data()
                if x[0] != x[-1]:
                    ln.set_data(np.append(x, x[0]), np.append(y, y[0]))
            return lines

        def set_varlabels(self, labels):
            self.set_thetagrids(np.degrees(theta), labels)

        def _gen_axes_patch(self):
            if frame == 'circle': return Circle((0.5, 0.5), 0.5)
            if frame == 'polygon':
                return RegularPolygon((0.5, 0.5), num_vars, radius=.5, edgecolor="k")
            raise ValueError(f"Unknown frame: {frame}")

        def _gen_axes_spines(self):
            if frame == 'circle': return super()._gen_axes_spines()
            if frame == 'polygon':
                spine = Spine(axes=self, spine_type='circle', path=Path.unit_regular_polygon(num_vars))
                spine.set_transform(Affine2D().scale(.5).translate(.5, .5) + self.transAxes)
                return {'polar': spine}
            raise ValueError(f"Unknown frame: {frame}")

    register_projection(RadarAxes)
    return theta

# ---------- multi-plot ----------
def plot_all_problems(runtime_data, problems=None, algorithms=None):
    """
    algorithms: list[str] specifying which algorithms to plot (legend uses this order).
                If None, uses palette order filtered by data.
    """
    if problems is None:
        problems = list(runtime_data.keys())

    # datasets from the first problem's first algo (keeps a consistent order)
    # datasets = list(next(iter(runtime_data[problems[0]].values())).keys())
    datasets = ["Wiki","Facebook", "Deezer","Slashdot", "Twitter", "DBLP",  "YouTube", "Skitter"][::-1]
    N = len(datasets)
    theta = radar_factory(N, frame='polygon')

    # figure with tighter spacing
    fig, axs = plt.subplots(
        figsize=(16.5, 5.6), nrows=1, ncols=len(problems),
        subplot_kw=dict(projection='radar')
    )
    axs = np.atleast_1d(axs)
    plt.subplots_adjust(wspace=0.08, left=0.03, right=0.97, top=0.88, bottom=0.18)

    # decide legend/plot order
    if algorithms is None:
        base_algos = list(runtime_data[problems[0]].keys())
        legend_labels = [a for a in palette if a in base_algos] + [a for a in base_algos if a not in palette]
    else:
        # keep only those present in ANY selected problem
        present = set()
        for p in problems:
            present |= set(runtime_data[p].keys())
        legend_labels = [a for a in algorithms if a in present]

    for ax, problem in zip(axs, problems):
        # center the gap at the bottom: π (bottom) + π/N (center-between spokes)
        offset = np.pi + (np.pi / N)
        ax.set_theta_offset(offset)

        for algo in legend_labels:
            if algo not in runtime_data[problem]:
                continue
            color = palette.get(algo, "#777777")
            vals = [np.log10(runtime_data[problem][algo][d]) for d in datasets]
            ax.plot(theta, vals, color=color, linewidth=2, label=algo)
            ax.fill(theta, vals, facecolor=color, alpha=0.20)

        ax.set_varlabels(datasets)
        angles_deg = (np.degrees(theta + offset)) % 360

        for lbl, ang in zip(ax.get_xticklabels(), angles_deg):
            # Radial outward orientation
            if 90 < ang < 270:  # left half, flip so text isn't upside down
                rotation = ang + 180
            else:
                rotation = ang
            lbl.set_rotation(rotation)
            lbl.set_horizontalalignment('center')
            lbl.set_verticalalignment('center')
            lbl.set_rotation_mode("anchor")

        # title + no y labels; fix useful radial range for your data
        ax.tick_params(pad=8) 
        ax.set_title(problem, size=13.5, pad=26, weight="bold")
        # ax.set_yticklabels(fontsize=14)   # hide y-axis labeling
        ax.set_ylim(-2, 4.7)     # log10 scale range (adjust if needed)

    # shared legend bottom-center following requested algorithms order
    # handles = [Line2D([0], [0], color=palette.get(a, "#777777"), lw=2)
    #        for a in legend_labels]
    handles = [Patch(facecolor=palette.get(a, "#777777"), edgecolor="none")
           for a in legend_labels]
    fig.legend(handles, legend_labels, loc="lower center",
            ncol=len(legend_labels), fontsize=16, frameon=False)
    plt.savefig("runtimes_radar.pdf", dpi=500, bbox_inches="tight")

    plt.show()

plot_all_problems(runtime_data,
                  ["Maximum Coverage", "Maximum Cut", "Influence Maximization"],
                  algorithms=["LLM2Prune","QuickPrune", "SS", "LeNSE"])
```

```python
problem = 'Maximum Coverage Weighted'
datasets = ['Facebook', 'DBLP', 'Deezer', 'Slashdot', 'Wiki', 'Twitter', 'YouTube']

print("\\begin{table}[h]")
print("\\centering")
print("\\begin{tabular}{lccc}")
print("\\toprule")
print("\\textbf{Dataset} & \\textbf{Ratio} & \\textbf{Size Reduction} & \\textbf{Combined Metric} ($\\\\times$) \\\\")
print("\\midrule")

for dataset in datasets:
    results = load_from_pickle(f'{problem}/{dataset}/results.pkl')
    ratio = results['Ratio'].iloc[0]
    size = results['Size Reduction'].iloc[0]
    combined = ratio * size
    print(f"{dataset} & {ratio:.4f} & {size:.4f} & {combined:.4f} \\\\")

print("\\bottomrule")
print("\\end{tabular}")
print("\\caption{Ratio, Size Reduction, and Combined Metric ($\\text{{Ratio}} \\times \\text{{Size Reduction}}$) across datasets}")
print("\\label{{tab:ratio_size_combined}}")
print("\\end{table}")
```

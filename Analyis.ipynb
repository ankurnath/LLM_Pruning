```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

for problem in [
                'Maximum Coverage',
                'Maximum Cut',
                'Influence Maximization',
                'Maximum Coverage Weighted',
                'Maximum Cut Weighted',
                'Influence Maximization Weighted'
                ]:
    datasets = ['Facebook', 'DBLP', 'Deezer', 'Slashdot', 'Wiki', 'Twitter', 'YouTube','Skitter']

    try:
        data_dict[problem]['LLM2Prune'] = {}
    except:
        data_dict[problem] = {}
        data_dict[problem]['LLM2Prune'] = {}


    for dataset in datasets:
        try:
            results = load_from_pickle(f'{problem}/{dataset}/results/{dataset}.pkl')

            ratio = float(results['Ratio'].iloc[0])
            size = float(results['Size Reduction'].iloc[0])
            combined = ratio * size

            data_dict[problem]['LLM2Prune'][dataset] = {
                "P_r": round(ratio, 4),
                "P_g": round(size, 4),
                "C": round(combined, 4)
            }
        except Exception as e:
            pass

def plot_algo_comparison(data_dict, problems, algos_to_plot, figsize=(12, 6)):
    for problem in problems:
        rows = []
        for algo, datasets in data_dict[problem].items():
            for dataset, vals in datasets.items():
                if vals["C"] is not None:
                    rows.append({"Algorithm": algo, "Dataset": dataset, "C": vals["C"]})
        df = pd.DataFrame(rows)
        df_sub = df[df["Algorithm"].isin(algos_to_plot)]
        plt.figure(figsize=figsize)
        sns.barplot(data=df_sub, x="Dataset", y="C", hue="Algorithm")
        plt.title(f"{problem} – Combined Metric (C)")
        plt.ylabel("Combined Metric (C)")
        plt.xticks(rotation=30)
        plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
        plt.tight_layout()
        plt.show()


plot_algo_comparison (data_dict= data_dict, problems = [
                                        'Maximum Coverage Weighted', 
                                        'Maximum Cut Weighted', 
                                        'Influence Maximization Weighted'
                                        ],
                      algos_to_plot=["LLM2Prune", "QuickPrune"])
```

```python
def compare_algorithms_avg(data_dict, problem, algo1, algo2):
    data = data_dict[problem]
    c1_vals, c2_vals = [], []
    for dataset in data[algo1]:
        c1, c2 = data[algo1][dataset]["C"], data[algo2][dataset]["C"]
        if c1 is not None and c2 is not None:
            c1_vals.append(c1)
            c2_vals.append(c2)
    avg1 = sum(c1_vals) / len(c1_vals) if c1_vals else None
    avg2 = sum(c2_vals) / len(c2_vals) if c2_vals else None
    if avg1 is None or avg2 is None: better = None
    elif avg1 > avg2: better = algo1
    elif avg2 > avg1: better = algo2
    else: better = "Tie"
    return {algo1: avg1, algo2: avg2, "Better": better}


compare_algorithms_avg(data_dict, problem = "Influence Maximization Weighted", algo1 = "QuickPrune", algo2 = "LLM2Prune")
```

```python
def count_algo_wins(data_dict, problem, algo1, algo2):
    data = data_dict[problem]
    wins = {algo1: 0, algo2: 0, "Tie": 0}
    for dataset in data[algo1]:
        c1, c2 = data[algo1][dataset]["C"], data[algo2][dataset]["C"]
        print(f'Dataset: {dataset}')
        print(f"{algo1} C:{c1}")
        print(f"{algo2} C:{c2}")
        if c1 is None or c2 is None:
            continue
        if c1 > c2:
            wins[algo1] += 1
        elif c2 > c1:
            wins[algo2] += 1
        else:
            wins["Tie"] += 1
    return wins
```

```python
def _latex_escape(s: str) -> str:
    # Escape LaTeX special chars in simple labels
    return (s.replace('\\', r'\textbackslash{}')
             .replace('_', r'\_')
             .replace('%', r'\%')
             .replace('&', r'\&')
             .replace('#', r'\#')
             .replace('{', r'\{')
             .replace('}', r'\}')
             .replace('$', r'\$'))

def _fmt4(x):
    if x is None:
        return r'--'
    try:
        return f"{float(x):.4f}"
    except Exception:
        return r'--'



```

```python
from matplotlib.table import table


def generate_latex_table_multi(data_dict, problems, algorithms, datasets=None, caption=None, label=None):
    """
    data_dict: nested dict (Problem -> Algorithm -> Dataset -> {P_r,P_g,C})
    problems: list[str]
    algorithms: list[str]
    datasets: optional list[str] (applied to each problem)
    caption/label: optional LaTeX caption/label.
    """

    # Column specification: left col + vertical bars separating algorithm blocks
    col_spec = "l|" + "|".join(["ccc"] * len(algorithms))

    # Top header
    top_header = [r"\textbf{Dataset}"]
    for idx, alg in enumerate(algorithms):
        alg_label = _latex_escape(alg).replace('*', r'\*')
        # add "|" only between algorithm blocks
        end_bar = "|" if idx < len(algorithms) - 1 else ""
        top_header.append(rf"\multicolumn{{3}}{{c{end_bar}}}{{\textbf{{{alg_label}}}}}")
    top_header = " & ".join(top_header) + r" \\"

    # Sub-header
    sub_header = [""]
    for _ in algorithms:
        sub_header += [r"\textbf{$P_r$}", r"\textbf{$P_g$}", r"\textbf{$C$}"]
    sub_header = " & ".join(sub_header) + r" \\"

    # light colors for ranks
    rank_colors = {1: "lightred", 2: "lightblue", 3: "lightgreen"}

    lines = []
    for prob in problems:
        if prob not in data_dict:
            continue
        # centered problem row
        lines.append(
            rf"\multicolumn{{{1+3*len(algorithms)}}}{{c}}{{\textbf{{{_latex_escape(prob)}}}}} \\"
        )
        algo_blocks = {a: data_dict[prob].get(a, {}) for a in algorithms}

        if datasets is None:
            seen = []
            for a in algorithms:
                for d in algo_blocks.get(a, {}).keys():
                    if d not in seen:
                        seen.append(d)
            prob_datasets = seen
        else:
            prob_datasets = datasets

        for ds in prob_datasets:
            # collect C values
            C_vals = []
            for alg in algorithms:
                m = algo_blocks.get(alg, {}).get(ds, {})
                try:
                    C_vals.append(float(m.get("C")))
                except:
                    C_vals.append(None)

            # rank values
            ranked = {}
            valid = [(i, c) for i, c in enumerate(C_vals) if c is not None]
            sorted_vals = sorted(valid, key=lambda x: x[1], reverse=True)
            for rank, (i, c) in enumerate(sorted_vals, start=1):
                if rank <= 3:
                    ranked[i] = rank

            row = [_latex_escape(ds)]
            for idx, alg in enumerate(algorithms):
                m = algo_blocks.get(alg, {}).get(ds, {})
                pr = _fmt4(m.get("P_r"))
                pg = _fmt4(m.get("P_g"))
                cval = _fmt4(m.get("C"))
                if idx in ranked:
                    color = rank_colors[ranked[idx]]
                    cval = rf"\textbf{{\textcolor{{{color}}}{{{cval}}}}}"
                row += [pr, pg, cval]
            lines.append(" & ".join(row) + r" \\")
        lines.append(r"\midrule")

    caption = caption or "Results across problems and datasets"
    label = label or "tab:multi_problem_algos"

    table = []
    table.append(r"\begin{table*}[h]")
    table.append(r"\centering")
    table.append(rf"\caption{{{caption}}}")
    table.append(rf"\label{{{label}}}")
    table.append(r"\resizebox{\textwidth}{!}{%")  # start resize
    table.append(rf"\begin{{tabular}}{{{col_spec}}}")
    table.append(r"\toprule")
    table.append(top_header)
    table.append(sub_header)
    table.append(r"\midrule")
    table.extend(lines)
    table.append(r"\bottomrule")
    table.append(r"\end{tabular}")
    table.append(r"}")  # end resize
    table.append(r"\end{table*}")


    return "\n".join(table)
```

```python
import pandas as pd 
import matplotlib.pyplot as plt
import numpy as np

from matplotlib.patches import Circle, RegularPolygon
from matplotlib.path import Path
from matplotlib.projections import register_projection
from matplotlib.projections.polar import PolarAxes
from matplotlib.spines import Spine
from matplotlib.transforms import Affine2D

# ---------- Global Style ----------
plt.rcParams["figure.dpi"] = 400       # high DPI
plt.rcParams["savefig.dpi"] = 300      # high-quality saves
plt.rcParams["font.size"] = 14         # base font size
plt.rcParams["axes.titlesize"] = 16    # title font
plt.rcParams["axes.labelsize"] = 14    # axis label font
plt.rcParams["legend.fontsize"] = 13   # legend font
plt.rcParams["xtick.labelsize"] = 15   # tick font
plt.rcParams["ytick.labelsize"] = 13

plt.rcParams["pdf.fonttype"] = 42    # ensures TrueType, not Type 3
plt.rcParams["ps.fonttype"] = 42
# plt.rcParams["font.family"] = "Helvetica"   # or "Arial"

# ---------- Radar Factory ----------
def radar_factory(num_vars, frame='circle', midpoint=True):
    base_theta = np.linspace(0, 2*np.pi, num_vars, endpoint=False)
    if midpoint:
        step = (2*np.pi) / num_vars
        theta = base_theta + step / 2
    else:
        theta = base_theta

    class RadarTransform(PolarAxes.PolarTransform):
        def transform_path_non_affine(self, path):
            if path._interpolation_steps > 1:
                path = path.interpolated(num_vars)
            return Path(self.transform(path.vertices), path.codes)

    class RadarAxes(PolarAxes):
        name = 'radar'
        PolarTransform = RadarTransform

        def __init__(self, *args, **kwargs):
            super().__init__(*args, **kwargs)
            self.set_theta_zero_location('N')

        def fill(self, *args, closed=True, **kwargs):
            return super().fill(closed=closed, *args, **kwargs)

        def plot(self, *args, **kwargs):
            lines = super().plot(*args, **kwargs)
            for line in lines:
                self._close_line(line)

        def _close_line(self, line):
            x, y = line.get_data()
            if x[0] != x[-1]:
                x = np.append(x, x[0])
                y = np.append(y, y[0])
                line.set_data(x, y)

        def set_varlabels(self, labels):
            self.set_thetagrids(np.degrees(theta), labels, fontsize=13)  # larger labels

        def _gen_axes_patch(self):
            if frame == 'circle':
                return Circle((0.5, 0.5), 0.5)
            elif frame == 'polygon':
                return RegularPolygon((0.5, 0.5), num_vars,
                                      radius=.5, edgecolor="k")
            else:
                raise ValueError(f"Unknown value for frame: {frame}")

        def _gen_axes_spines(self):
            if frame == 'circle':
                return super()._gen_axes_spines()
            elif frame == 'polygon':
                spine = Spine(self, 'circle',
                              Path.unit_regular_polygon(num_vars))
                spine.set_transform(Affine2D().scale(.5).translate(.5, .5)
                                    + self.transAxes)
                return {'polar': spine}
            else:
                raise ValueError(f"Unknown value for frame: {frame}")

    register_projection(RadarAxes)
    return theta


# ---------- Data Prep ----------
rows = []
for problem, algos in runtime_data.items():
    for algo, datasets in algos.items():
        for dataset, val in datasets.items():
            if val is not None:
                rows.append({
                    "Problem": problem,
                    "Algorithm": algo,
                    "Dataset": dataset,
                    "Runtime": float(val)
                })
df = pd.DataFrame(rows)

algos_keep = ["SS", "QuickPrune", "LLM2Prune", "LeNSE"]
df = df[df["Algorithm"].isin(algos_keep)]
df["Score"] = np.log10(df["Runtime"])   # absolute log runtime


# ---------- Plot ----------
colors = {
    "SS": "#6A5ACD",
    "QuickPrune": "#FF6347",
    "LLM2Prune": "#FFB000",
    "LeNSE": "#20B2AA",
}

problems = df["Problem"].unique()
ncols = len(problems)

fig, axs = plt.subplots(figsize=(5*ncols+3, 6),
                        nrows=1, ncols=ncols,
                        subplot_kw=dict(projection="radar"))

if ncols == 1:
    axs = [axs]

for ax, problem in zip(axs, problems):
    sub = df[df["Problem"] == problem]
    categories = sub["Dataset"].unique()
    N = len(categories)
    theta = radar_factory(N, frame="polygon", midpoint=True)

    # Plot each algorithm
    for algo, color in colors.items():
        values = (
            sub[sub["Algorithm"] == algo]
            .set_index("Dataset")
            .reindex(categories)["Score"]
            .fillna(0)
            .tolist()
        )
        ax.plot(theta, values, color=color, linewidth=2, label=algo)
        ax.fill(theta, values, facecolor=color, alpha=0.25)

    # labels + limits
    ax.set_ylim(-1.5, 4.6)
    ax.set_varlabels(categories)
    ax.set_yticklabels([])
    ax.set_title(f"{problem}", size=18, pad=25, weight="bold")

# ---------- Shared Legend ----------
handles, labels = axs[0].get_legend_handles_labels()
# fig.legend(handles, labels,
#            loc="lower center",
#            ncol=len(algos_keep),
#            frameon=False,
#            fontsize=14)

# fig.legend(handles, labels, loc="lower center", ncol=len(algos_keep),
#            frameon=True, fancybox=True, shadow=False,
#            framealpha=0.9, facecolor="white", edgecolor="0.8")

from matplotlib.patches import Patch

# ---------- Shared Legend ----------
handles = [Patch(facecolor=colors[algo], edgecolor="none", label=algo) 
           for algo in algos_keep]

fig.legend(handles,
           [h.get_label() for h in handles],
           loc="lower center",
           ncol=len(algos_keep),
           frameon=False,
           fontsize=16)


plt.subplots_adjust(wspace=0.5, bottom=0.15)  
plt.tight_layout(rect=[0, 0.12, 1, 1])
plt.savefig("radar_runtime_comparison.pdf", bbox_inches="tight",dpi=500)
plt.show()
```
